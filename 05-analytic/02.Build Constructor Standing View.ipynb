```sql
create or replace view  formula1_catalog.gold.v_constructor_standings
as 
with constructor_season_summry as
(
select r.season,
       c.constructor_id,
       c.constructor_name,
       c.nationality,
       count(*) as race_start,
       SUM(r.race_points) as total_points,
       count_if(isWin) as number_of_wins,
       count_if(isPodium) as number_of_podiums
 from formula1_catalog.gold.fact_session_results r
join formula1_catalog.gold.dim_constructors c
on r.constructor_id = c.constructor_id
group by r.season,
       c.constructor_id,
       c.constructor_name,
       c.nationality
)
SELECT season,
       constructor_id,
       constructor_name,
       nationality,
       rank() over (partition by season order by total_points desc,number_of_wins desc) as satnding,
       race_start,
       total_points,
       number_of_wins,
       number_of_podiums 
FROM constructor_season_summry
```
