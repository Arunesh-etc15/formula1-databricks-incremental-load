#Set upenvironment for formula1 project
1.create external location databricksstoragea

2.create catalog formula1

```sql
%fs ls 'abfss://formula1-incr@databricksstoragea.dfs.core.windows.net/landing'
```

```sql
CREATE EXTERNAL LOCATION if not EXISTS databricksstoragea_formula1_incr
URL 'abfss://formula1-incr@databricksstoragea.dfs.core.windows.net/'
WITH (STORAGE CREDENTIAL data_bricks_ex_sc)
comment "External location for databricksstoragea formula1 incremnetal container"
```

```sql
show catalogs;
```

```sql
CREATE CATALOG formula1_incr_catalog
MANAGED LOCATION 'abfss://formula1-incr@databricksstoragea.dfs.core.windows.net/'
comment "Main catalog for formula1 incremnetal";
```

```sql
CREATE SCHEMA if not EXISTS formula1_incr_catalog.landing;
CREATE SCHEMA if not EXISTS formula1_incr_catalog.bronze 
  MANAGED LOCATION 'abfss://formula1-incr@databricksstoragea.dfs.core.windows.net/bronze' ;
CREATE SCHEMA if not EXISTS formula1_incr_catalog.silver 
  MANAGED LOCATION 'abfss://formula1-incr@databricksstoragea.dfs.core.windows.net/silver' ;
CREATE SCHEMA if not EXISTS formula1_incr_catalog.gold 
  MANAGED LOCATION 'abfss://formula1-incr@databricksstoragea.dfs.core.windows.net/gold' 
```
